```python
class SVM:

  #initiating hyper parameters
  def __init__(self,lr,w,l):
    #lr:Learning Rate, w:Weight ,l:lamda parameter
    self.lr=lr
    self.w=w
    self.l=l

  def fit(self,X,Y):
    self.X=X
    self.Y=Y
    self.m,self.n=X.shape # m:rows,n:columns
    self.b=0 #b:bias
    self.w=np.zeros(self.n)

    # implementing gradient decent algorithm for optimization
    for i in range(self.lr):
      update_weights()

  def update_weights(self):
    y_label=np.where(self.Y<=0,-1,1)

    for index,x_i in enumerate (self.X):
      condition=y_label [index]*(np.dot(x_i,self.w)-self.b)>=1
      if(condition==true):
        dw=2*self.w*self.l
        db=0
      else:
        dw=2*self.w*self.l-np.dot(x_i,y_label[index])
        db=y_label[index]


```
