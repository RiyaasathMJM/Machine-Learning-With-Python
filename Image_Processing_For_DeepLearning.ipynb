RGB to Grayscale

```python
#Importing opencv library
import cv2
```

```python
img=cv2.imread('/content/car.png')
```

```python
type(img)
```

```python
img.shape
```

```python
grayscale_img=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
```

```python
type(grayscale_img)
```

```python
grayscale_img.shape
```

```python
from google.colab.patches import cv2_imshow
cv2_imshow(grayscale_img)
```
